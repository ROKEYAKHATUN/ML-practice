```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(1)
df=pd.DataFrame(np.random.rand(1000,4),columns=['A','B','C','D'])
df.info()
```

```python
df.head()
```

```python
df.plot.scatter(x='C', y='D', figsize=(8,3),color='g')
```

```python

df['D'].hist(figsize=(6,3),grid=False)
plt.ylabel('Frequency')
```

```python
plt.style.use('bmh')
df['D'].hist(bins=50,figsize=(6,3))
plt.ylabel('Frequency')
```

```python
df.plot.scatter(x='A',y='B', s=df['D']*80,color='g',figsize=(6,3))
```

```python
df.plot.hexbin(x='A', y='B', cmap='coolwarm', gridsize=20, figsize=(6,3))
```

```python
color=dict(boxes='DarkGreen', whiskers='DarkOrange',
           medians='DarkBlue', caps='Gray')
df.plot.box(color=color,sym='r+')
```
