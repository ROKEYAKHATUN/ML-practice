```python
import numpy as np
x=np.arange(0,10)
y = x*2
z = x**2
```

```python
import matplotlib.pyplot as plt
fig1=plt.figure(figsize=(6,5))
axes=fig1.add_axes([0,0,1,1])
axes.plot(x, y,'g',label="x**2")
axes.set_xlabel('x_label')
axes.set_ylabel('y_label')
axes.set_title('scatter plot')
axes.set_xticks(np.arange(0, 9, 2))
axes.set_yticks(np.arange(0, 18.1, 2.5))
axes.legend(loc=4)
plt.show()

```

```python

fig3 = plt.figure(figsize=(10,5))
axes_main = fig3.add_axes([0,0,1,1])
axes_inset = fig3.add_axes([0.1, 0.6, 0.2, 0.2])

# Main figure
axes_main.plot(x, z, 'r',label="x**2")
axes_main.set_xlabel('X_label_main')
axes_main.set_ylabel('Y_label_main')
axes_main.set_title('Title_main')
axes_main.legend(loc=4)

# Inset figure
axes_inset.plot(x, z, 'b',label="x**2")
axes_inset.set_xlabel('X_inset')
axes_inset.set_ylabel('Y_inset')
axes_inset.set_title('Inset')
axes_inset.set_xlim(1,2)
axes_inset.set_ylim(0,5)
axes_inset.set_xticks(np.arange(1, 2.1, 0.5))
axes_inset.set_yticks(np.arange(0, 5.1, 2.5))
axes_inset.legend()
plt.show()
```

```python
plt.subplot(1,3,1)
plt.plot(x,y,'b')

plt.subplot(1,3,2)
plt.plot(x,y,'b')

plt.subplot(1,3,3)
plt.plot(x,y,'b')
plt.tight_layout()
```
