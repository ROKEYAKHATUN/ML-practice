```python
import pandas as pd
```

```python
from sklearn import datasets
iris=datasets.load_iris()
iris
```

```python
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df
```

```python
df['flower']=iris.target
df
```

```python
x=df.drop('flower',axis=1)
y=df.flower
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.svm import SVC
model=SVC(kernel='rbf', C=1, gamma='auto')
model.fit(x_train,y_train)
```

```python
model.score(x_test, y_test)
```

```python
from sklearn.model_selection import cross_val_score
sv1=cross_val_score(SVC(kernel='rbf', C=1, gamma='auto'),x,y,cv=5)
sv1
```

```python
sv1.mean()
```

```python
sv2=cross_val_score(SVC(kernel='poly',C=1,gamma='auto'),x, y, cv=5)
sv2.mean()
```

```python
sv3=cross_val_score(SVC(kernel='rbf',C=20,gamma='auto'),x, y, cv=5)
sv3.mean()
```

```python
from sklearn.tree import DecisionTreeClassifier
sv4 = cross_val_score(DecisionTreeClassifier(), x, y, cv=5)
sv4.mean()
```

```python
from sklearn.naive_bayes import GaussianNB
sv5 = cross_val_score(GaussianNB(), x, y, cv=5)
sv5.mean()
```

```python
from sklearn.neighbors import KNeighborsClassifier
sv6 = cross_val_score(KNeighborsClassifier(), x, y, cv=5)
sv6.mean()
```

```python
from sklearn.model_selection import GridSearchCV
clf = GridSearchCV(SVC(gamma='auto'), {
    'C': [1,10,20,30],
    'kernel': ['rbf','linear']
}, cv=5)
clf.fit(x, y)
clf.cv_results_
```

```python
df=pd.DataFrame(clf.cv_results_)
df
```

```python
df[['param_C','param_kernel','mean_test_score']]
```
