```python
df=pd.read_csv('Cancer.csv')
df
```

```python
df.info()
```

```python
x = df.drop('diagnosis', axis = 1)
y = df['diagnosis']
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3, random_state = 2)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
clf=GridSearchCV(SVC(gamma='auto'),
 {'C':[1,10,20,30],
  'kernel':['rbf','linear']},
  cv=5)
clf.fit(x_train,y_train)
clf.cv_results_
```

```python
df =pd.DataFrame(clf.cv_results_)
df
```

```python
clf.best_params_
```

```python
clf.best_score_
```

```python
svm_model = SVC(kernel='linear',C=30,gamma='auto')
svm_model.fit(x_train,y_train)
```

```python
svm_model.score(x_test, y_test)
```

```python
svm_model.predict([[1.740, 1.91, 10.12, 585.0, 0.07944, 0.06376, 1.02881, 0.01329, 0.1473, 2.05580]])
```

```python
predictions = svm_model.predict(x_test)
predictions
```

```python
from sklearn.preprocessing import MinMaxScaler

MMS = MinMaxScaler()

x_after_min_max_scaler = MMS.fit_transform(x_train)

x_after_min_max_scaler
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_after_min_max_scaler, y_train)
```
